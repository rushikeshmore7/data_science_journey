##  Comparison Operators 

### 1.Compare two numbers using ==.


```python
a=22
b=21
equal=a==b
print(equal)
```

### 2 Compare using !=.


```python
a=22
b=21
equal= a!=b
print(equal)
```

### 3 Compare using >.


```python
a=2
b=4
greater=a>b
print(greater)
```

### 4 Compare using <.


```python
a=2
b=3
less=a<b
print(less)
```

### 5 Compare using >=.


```python
a=12
b=12
gequal=a>=b
print(gequal)
```

### 6 Compare using <=.


```python
a=12
b=24
print(a<=b)
```

### 7 Check whether two strings are equal.


```python
a="1","2","4"
b="1","3"

print(a==b)
```

### 8 Compare two lists using ==.Compare two tuples.


```python
lst=[1,2,3]
tup=(1,2,3)
print(lst == tup)
```

### 9 Compare two floats.


```python
a=7.4
b=7.4
print(a!=b)
print(a<=b)
print(a>=b)
print(a==b)
print(a>b)
print(a<b)
```

### 10 Find the larger of two numbers using >.


```python
a=10
b=20
if a>b:
    print("a is greater:",a)
else:
    print("b is greater than a:",b)
```

### 11 Check whether a number is positive.


```python
a =1

if a>0:
    print("positive number :",a)

else:
    print("negative number:",a)
```

### 12 Check whether a number is negative.


```python
a =-1

if a<0:
    print("negative number :",a)

else:
    print("positive number:",a)
```
