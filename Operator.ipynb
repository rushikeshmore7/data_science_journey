### 1.Use += to increase a number by 10.


```python
a=10
a+=10
print(a)
```

### 2.Use -= to decrease a number by 5.


```python
a=10
a-=5
print(a)
```

### 3.Use *= to double a number.


```python
a=4
a*=2
print(a)
```
